```python
# step 1 : Create and print a list
shopping = ["rice", "eggs", "milk"] 
print("Start :", shopping)
```

```python
# step 2 : add a bread to the end and water to the front
shopping.append("bread") 
shopping.insert(0, "water") 
print("Added :", shopping)
```

```python
# Step 3 : remove egg and the last item in the list 
shopping.remove("eggs") 
last = shopping.pop() 
print("Removed :", last) 
print("Now :", shopping)
```

```python
# Step 4 : slicing method with list
print("First 2 :", shopping[:2]) 
print("Last 2 :", shopping[-2:]) 
print("Reversed:", shopping[::-1])
```

```python
# Step 5 : check if milk in list named shopping 
print("Is milk in the list?", "milk" in shopping) 
for i, item in enumerate(shopping, start = 1): 
    print(i, item)
```
