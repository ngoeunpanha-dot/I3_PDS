```python
#step3
student = ( "dara",20,3.45),("Sophea",21,3.92),("Vibol",19,3.10),("Malis",22,2.78)
for name,age,gpa in student:
    print(name,age,gpa)
```

```python
#step4
rank = sorted(student,key=lambda s: s[2],reverse=True) # sortby gpa and use key tell sorted what value to use then lamda s:s[2]means sort by the 3rd value of each tuple 
print("Best student:",rank[0][0])
```
